## Explorar el contenido descargado

Lista el contenido de la carpeta de "Video of vehicle" una vez que termine la celda de descarga de arriba.

```python
import os

for root, dirs, files in os.walk(jimenavargasvega_video_of_vehicle_path):
    nivel = root.replace(jimenavargasvega_video_of_vehicle_path, "").count(os.sep)
    print("  " * nivel + os.path.basename(root) + "/")
    for f in files[:10]:
        print("  " * (nivel + 1) + f)
    if len(files) > 10:
        print("  " * (nivel + 1) + f"... y {len(files) - 10} archivos mas")
```
